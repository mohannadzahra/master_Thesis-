```python
import pandas as pd
import os
```

```python
# Load the data from the text file
#file_path = 'subset_1_of_TSX1_SAR_SSC_SL_S_SRA_20231230T074146_20231230T074147_Stack_ifg_dinsar_ML_10_flt_TC_geometry_Mask.tx'
#file_path = r"C:\Users\Mohannad Zahra\iceland\Grindavik_Svartsengi_data\The second episode from 2023-12-19 to 2024-01-13\track_49\coh\subset_2_of_TSX1_SAR_SSC_SL_S_SRA_20231230T074146_20231230T074147_Stack_ifg_dinsar_ML_10_flt_TC_geometry_Mask.txt"

file_path = r"C:\Users\Mohannad Zahra\iceland\Grindavik_Svartsengi_data\The fifth episode from 2024-03-17 to 2024-05-28\track_102\coh_2024_05_13\subset_0_of_TDX1_SAR_SSC_SL_S_SRA_20240410T191454_20240410T191456_Stack_ifg_dinsar_ML_10_flt_TC_geometry_Mask.txt"
# Read the file using space as the delimiter
#df = pd.read_csv(file_path, delim_whitespace=True)
data = pd.read_csv(file_path, sep="\t")
```

```python
#df
data
```

```python
# Calculate the mean of the 'mean_coherence' column
#mean_coh = data['coh_HH_30Dec2023_10Jan2024'].mean()
mean_coh = data['coh_HH_10Apr2024_13May2024'].mean()

# Define the product name and Date (from the file details)
#product_name = 'Mean_subset_2_coh_HH_30Dec2023_10Jan2024'
product_name = 'Mean_subset_0_coh_HH_10Apr2024_13May2024'
Date = '10Apr2024_13May2024'

#Date = '30Dec2023_10Jan20240
```

```python
# Create a DataFrame with the result
output = pd.DataFrame({
    'Product_Name': [product_name],
    'Date Range': [Date],
    'Mean_coherence': [mean_coh]
})
```

```python
# Save the result to a CSV file
output_file_path = r"C:\Users\Mohannad Zahra\iceland\Grindavik_Svartsengi_data\coh\mean_coh_0_output_new.csv"
#output_file_path = r"C:\Users\Mohannad Zahra\iceland\Grindavik_Svartsengi_data\coh\mean_coh_1_output_new.csv"
#output_file_path = r"C:\Users\Mohannad Zahra\iceland\Grindavik_Svartsengi_data\coh\mean_coh_2_output_new.csv"

# Append the DataFrame to the CSV file (if the file exists, it will append, otherwise it will create it)
header = not os.path.exists(output_file_path)  # Write header if file doesn't exist

# Save the DataFrame to the local CSV file
output.to_csv(output_file_path, mode='a', header=header, index=False)

# Save the DataFrame to the local CSV file
#output.to_csv(output_file_path, index=False)

print(f"Mean coherence calculated and saved to {output_file_path}")
```
